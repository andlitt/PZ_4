# Создание Series

```python
import numpy as np
import pandas as pd

s1 = pd.Series([1,2,3,4,5],['a','b','c','d','e'])
print(s1)
```

#  Создание Series из ndarray массива из NumPy

```python
ndarr = np.array([1,2,3,4,5])
type(ndarr)
s3 = pd.Series(ndarr,['a','b','c','d','e'])
print(s3)
```

# Создание Series из словаря (dict)

```python
d = {'a':1,'b':2,'c':3}
s4 = pd.Series(d)
print(s4)
```

# Создание Series с использованием константы

# Позиционная индексация (iloc)

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])

print(s.iloc[0])
print(s.iloc[2])
print(s.iloc[-1])
```

#Меточная индексация (loc)

```python
s = pd.Series([10,20,30,40,50], index=['a','b','c','d','e'])
print(s.loc['a'])
print(s.loc['c'])
print(s.loc['e'])
```

# Использование логической индексации для фильтрации данных в Series

```python
s = pd.Series([10,25,8,30,15], index=['a','b','c','d','e'])

# Создаём Series с числовыми значениями
filtered_s = s[s>10] # Фильтруем значения больше 10

print(filtered_s)
```

#Фильтрация по значению индекса

```python
filtered_s = s[s.index.isin(['b', 'd'])]
print(filtered_s)
```

# Операции над Series**

# Арифметические операции с числом

```python
s = pd.Series([10,20,30,40,50])
s_multiplied = s * 2
print(s_multiplied)
```

# Арифметические операции между двумя Series

```python
s1 = pd.Series([1,2,3,4,5], index=['a','b','c','d','e'])
s2 = pd.Series([10,20,30,40,50], index=['a','b','c','d','e'])

# Поэлементное сложение двух Series
s_sum = s1+s2
print(s_sum)
```

# Работа с несовпадающими индексами

```python
s1 = pd.Series([1,2,3,4,5], index=['a','b','c','d','e'])
s3 = pd.Series([100,200,300], index=['a','b','f'])

# Сложение Series с несовпадающими индексами
s_result = s1 + s3
print(s_result)
```

# Использование методов .add(), .sub(), .mul(), .div()

```python
# Заменяем NaN на 0 при сложении
s1 = pd.Series([1,2,3,4,5], index=['a','b','c','d','e'])
s_result = s1.add(s3, fill_value=0)
print(s_result)
```

#Применение функций .apply() для Series

```python
s = pd.Series([1,2,3,4,5])
s_squared = s.apply(lambda x: x**2) # Возводим в квадрат каждый элемент
print(s_squared)
```

# Использование .apply() с обычной функцией

```python
s = pd.Series([1,2,3,4,5])

def custom_function(x): # Функция, преобразующая числа в строки
  return f"Value: {x}"

s_transformed = s.apply(custom_function)
print(s_transformed)
```

# Создание временного ряда в Series

```python
dates = pd.date_range(start='2024-03-01', periods=5, freq='D')
s = pd.Series([100,105,102,98,110], index=dates)
print(s)
```

# Доступ к данным по дате

```python
print(s['2024-03-03'])
print("\n",s['2024-03-02':'2024-03-04']) # Фильтрация по диапазону дат
```

# Создание временного ряда с разной частотой

```python
# Создание временного ряда с интервалом в 1 час
dates = pd.date_range(start='2024-03-01', periods=5, freq='h')
s = pd.Series([10,15,12,8,20], index=dates)
print(s)
```

# Перевод индекса в формат Datetime

```python
s = pd.Series([100,200,150], index=['2024-03-01','2024-03-02','2024-03-03'])
s_index = pd.to_datetime(s.index)
print(s)
```

# Расчет скользящего среднего

```python
s = pd.Series([10,20,30,40,50,60,70], index=pd.date_range('2024-03-01', periods=7,freq="D"))
s_ma = s.rolling(window=3).mean()
print(s_ma)
# Первые два значения NaN, потому что окно (window=3) не содержит достаточно элементов для вычисления среднего
```

# Скользящее среднее с минимальным количеством значений

```python
s_ma2 = s.rolling(window=3, min_periods=1).mean()
print(s_ma2)
# Теперь даже первые элементы  получают усредненные значения
```

# Построение графика скользящего среднего

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,4))
plt.plot(s, label="Исходные данные", marker='o')
plt.plot(s.rolling(window=3).mean(), label="Скользящее среднее (3)", linestyle='--', color='red')
plt.xlabel('Дата')
plt.ylabel('Значение')
plt.title('Скользящее среднее в Series')
plt.legend()
plt.grid()
plt.show()
```

# Применение .pct_change() к временным рядам

```python
# Генерируем данные о цене акции
dates = pd.date_range(start="2024-01-01", periods=30, freq="D")
prices = pd.Series(np.random.randint(90, 110, size=30), index=dates)

# Вычисляем процентный прирост
returns = prices.pct_change() * 100

# Построим график
plt.figure(figsize=(10, 5))
plt.plot(prices, label="Цена акции", marker="o")
plt.ylabel("Цена")
plt.xlabel("Дата")
plt.title("График цены акции")

plt.figure(figsize=(10, 5))
plt.bar(returns.index, returns, color="red", alpha=0.7, label="Процентное изменение")
plt.axhline(0, color="black", linestyle="--")
plt.ylabel("Изменение (%)")
plt.xlabel("Дата")
plt.title("Процентное изменение цены акции")
plt.legend()
plt.show()
```
