6. Проверка пропущенных значений

Создайте Series с данными [1.2, np.nan, 3.4, np.nan, 5.6, 6.8] . Напишите код, который проверяет, есть ли в Series пропущенные значения ( NaN ), и выведите индексы таких элементов.

```python
data = pd.Series([1.2, np.nan, 3.4, np.nan, 5.6, 6.8])

has_nans = data.isna().any()
print("Есть ли пропущенные значения (NaN)?", has_nans)

nan_indices = data[data.isna()].index.tolist()
print("Индексы пропущенных значений:", nan_indices)
```

7. Заполнение пропущенных значений

8. Арифметические операции с Series
   
Создайте два Series :
s1 = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
s2 = pd.Series([5, 15, 25, 35], index=['b', 'c', 'd', 'e']) Выполните сложение s1 + s2 . Объясните, почему в результате появляются NaN , и замените их на 0 .

```python
s1 = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
s2 = pd.Series([5, 15, 25, 35], index=['b', 'c', 'd', 'e'])

result = s1 + s2
result_filled = result.fillna(0)

print("Результат сложения (с NaN):\n", result)
print("\nРезультат после замены NaN на 0:\n", result_filled)

# При операциях между Series pandas складывает только элементы с совпадающими индексами.
# Индексы 'a' (есть только в s1) и 'e' (есть только в s2) не имеют пары, значит pandas ставит NaN.
```

10. Основные статистические методы
    
Создайте Series из 20 случайных чисел от 50 до 150 (используйте np.random.randint ). Найдите сумму, среднее, минимальное и максимальное значение. Выведите также стандартное отклонение.

```python
random_series = pd.Series(np.random.randint(50, 151, size=20))

stats = {
    "Сумма": random_series.sum(),
    "Среднее": random_series.mean(),
    "Минимум": random_series.min(),
    "Максимум": random_series.max(),
    "Стандартное отклонение": random_series.std()
}

print("Series:")
print(random_series)

print("\nСтатистические показатели:")
for key, value in stats.items():
    print(f"{key}: {value:.2f}" if isinstance(value, float) else f"{key}: {value}")
```

12. Проверка уникальности индексов
    
Создайте Series синдексами ['A', 'B', 'A', 'C', 'D', 'B'] и значениями [10, 20, 30, 40, 50, 60] . Проверьте, являются ли индексы уникальными. Если нет, сгруппируйте повторяющиеся индексы и сложите их значения.

```python
data = pd.Series([10, 20, 30, 40, 50, 60], 
                 index=['A', 'B', 'A', 'C', 'D', 'B'])

is_unique = data.index.is_unique
print(f"Индексы уникальные? {is_unique}")

# Если индексы не уникальны, группируем и суммируем
if not is_unique:
    grouped_data = data.groupby(level=0).sum()
    print("\nРезультат после группировки:")
    print(grouped_data)
```

15. Построение графика на основе Series
    
Создайте Series , где индексами будут даты с 1 по 30 марта 2024 года, а значениями - случайные числа от 50 до 150. Постройте график значений с помощью matplotlib . Добавьте заголовок подписи осей и сетку.

```python
import matplotlib.pyplot as plt

# Создаем Series с датами и случайными значениями
dates = pd.date_range(start='2024-03-01', end='2024-03-30', freq='D')
values = np.random.randint(50, 151, size=len(dates))
time_series = pd.Series(values, index=dates, name='Случайные значения')


plt.figure(figsize=(12, 6))  
time_series.plot(color='blue', linewidth=2, marker='o', markersize=5)
plt.title('Случайные значения с 1 по 30 марта 2024 года', fontsize=14, pad=20)
plt.xlabel('Дата', fontsize=12)
plt.ylabel('Значения', fontsize=12)
plt.grid(True, linestyle='--')  
```

5. Преобразование и анализ погодных данных
   
Создайте Series , где индексами будут даты с 10 по 20 марта 2024 года, а значениями - случайные температуры от -5 до +10 градусов. Выделите данные за 15-17 марта, преобразуйте индекс в DatetimeIndex , а затем постройте график температуры.

```python
dates = pd.date_range(start='2024-03-10', end='2024-03-20', freq='D')
temperatures = np.random.randint(-5, 11, size=len(dates))  # Температуры от -5 до +10
temp_series = pd.Series(temperatures, index=dates, name='Температура')

# Выделяем данные за 15-17 марта
selected_data = temp_series['2024-03-15':'2024-03-17']

# Преобразуем индекс в DatetimeIndex (хотя date_range уже создает его)
selected_data.index = pd.to_datetime(selected_data.index)

# Строим график
plt.figure(figsize=(10, 5))
selected_data.plot(
    style='o-', 
    color='red',
    linewidth=2,
    markersize=8,
    markerfacecolor='white'
)


plt.title('Температура воздуха 15-17 марта 2024', fontsize=14, pad=15)
plt.xlabel('Дата', fontsize=12)
plt.ylabel('Температура (°C)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.7)
plt.axhline(0, color='black', alpha=0.35)  
plt.yticks(np.arange(-5, 11, 2))  
```
